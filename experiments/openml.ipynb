```python
import openml
import pandas as pd
from bonXAI.core.tabular_preprocessor import TabularPreprocessor

# Load from OpenML ID 1461 (task 7592)
d = openml.datasets.get_dataset("bank-marketing", version=1)
X, y, cat_mask, attr_names = d.get_data(target=d.default_target_attribute)
X = pd.DataFrame(X, columns=attr_names[:-1])
y = pd.Series(y, name=d.default_target_attribute)
df = pd.concat([X, y], axis=1)

# add id like columns to test
df["id_like_int"] = range(len(X))
df['id_like_float'] = df['id_like_int'].astype(float) 

# add a duplicate to test cleaner
df = pd.concat([df, df.iloc[[0]]], ignore_index=True)

X, y = df.drop(columns=[d.default_target_attribute]), df[d.default_target_attribute]

# fit + transform on the whole set
prep = TabularPreprocessor(
    task_type="classification",
    id_like_threshold=0.99,
    scale_all_numeric_after_encoding=True,
    scale_target_in_regression=False,
    random_state=0,
)
X_clean, y_enc = prep.fit_transform(X, y)

print("Original shape:", X.shape)
print("Processed shape:", X_clean.shape)
print("Task type:", prep.get_report().task_type)
print("Dropped constant:", prep.get_report().dropped_constant)
print("Dropped ID-like:", prep.get_report().dropped_id_like)
```

```python
print(prep.get_report())
print(y_enc)
X_clean.head(5)
```

# CTR23

```python
import openml
import pandas as pd
from bonXAI.core.utils import CTR23, CC18
from bonXAI.core.loaders import load_openml_data_and_model
from sklearn.metrics import mean_squared_error

data = []
for dataset_id in CTR23:
    try:
        X_train, y_train, X_test, y_test, model, prep = load_openml_data_and_model(dataset_id=dataset_id, model_name="ann", task_type="classification")
        dataset = openml.datasets.get_dataset(dataset_id)
        X, y, _, _ = dataset.get_data(target=dataset.default_target_attribute, dataset_format="dataframe")
        mse = mean_squared_error(y_test, model.predict(X_test))
        print(mse)
        data.append({
            'dataset_id': dataset_id,
            'name': dataset.name,
            'n_samples_train': X_train.shape[0],
            'n_samples_test': X_test.shape[0],
            'n_features_before_preprocessing': X.shape[1],
            'n_features_after_preprocessing': X_train.shape[1], 
            'mse': mse
        })
    except Exception as e:
        print(f"Error fetching dataset {dataset_id}: {e}")
        data.append({
            'dataset_id': dataset_id,
            'name': None,
            'n_samples_train': None,
            'n_samples_test': None,
            'n_features_before_preprocessing': None,
            'n_features_after_preprocessing': None, 
            'mse': None
        })

# Create DataFrame
df = pd.DataFrame(data)
print(df)
```

```python
# save to csv
df.to_csv("CTR23_datasets_summary.csv", index=False)
```

# CC18

```python
from sklearn.metrics import accuracy_score
import openml
import pandas as pd
from bonXAI.core.utils import CTR23, CC18
from bonXAI.core.loaders import load_openml_data_and_model

data = []
for dataset_id in CC18:
    try:
        X_train, y_train, X_test, y_test, model, prep = load_openml_data_and_model(dataset_id=dataset_id, model_name="ann", task_type="classification")
        dataset = openml.datasets.get_dataset(dataset_id)
        X, y, _, _ = dataset.get_data(target=dataset.default_target_attribute, dataset_format="dataframe")
        # calculate accuracy on test set
        y_pred = model.predict(X_test)
        accuracy = accuracy_score(y_test, y_pred)
        print("Accuracy:", accuracy)
        data.append({
            'dataset_id': dataset_id,
            'name': dataset.name,
            'n_samples_train': X_train.shape[0],
            'n_samples_test': X_test.shape[0],
            'n_features_before_preprocessing': X.shape[1],
            'n_features_after_preprocessing': X_train.shape[1], 
            'accuracy': accuracy
        })
        
    except Exception as e:
        print(f"Error fetching dataset {dataset_id}: {e}")
        data.append({
            'dataset_id': dataset_id,
            'name': None,
            'n_samples_train': None,
            'n_samples_test': None,
            'n_features_before_preprocessing': None,
            'n_features_after_preprocessing': None, 
            'accuracy': None
        })

# Create DataFrame
df = pd.DataFrame(data)
print(df.head())
```

```python
df.to_csv("CC18_datasets_summary.csv", index=False)
```
