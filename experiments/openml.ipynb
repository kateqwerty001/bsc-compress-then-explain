```python
import openml
import pandas as pd
from bonXAI.core.tabular_preprocessor import TabularPreprocessor

# Load from OpenML ID 1461 (task 7592)
d = openml.datasets.get_dataset("bank-marketing", version=1)
X, y, cat_mask, attr_names = d.get_data(target=d.default_target_attribute)
X = pd.DataFrame(X, columns=attr_names[:-1])
y = pd.Series(y, name=d.default_target_attribute)
df = pd.concat([X, y], axis=1)

# add id like columns to test
df["id_like_int"] = range(len(X))
df['id_like_float'] = df['id_like_int'].astype(float) 

# add a duplicate to test cleaner
df = pd.concat([df, df.iloc[[0]]], ignore_index=True)

X, y = df.drop(columns=[d.default_target_attribute]), df[d.default_target_attribute]

# fit + transform on the whole set
prep = TabularPreprocessor(
    task_type="classification",
    id_like_threshold=0.99,
    scale_all_numeric_after_encoding=True,
    scale_target_in_regression=False,
    random_state=0,
)
X_clean, y_enc = prep.fit_transform(X, y)

print("Original shape:", X.shape)
print("Processed shape:", X_clean.shape)
print("Task type:", prep.get_report().task_type)
print("Dropped constant:", prep.get_report().dropped_constant)
print("Dropped ID-like:", prep.get_report().dropped_id_like)
```

# CTR23

```python
import openml
import pandas as pd
from bonXAI.core.utils import CTR23
from bonXAI.core.data_loader import DataLoader
from sklearn.metrics import mean_squared_error

loader = DataLoader()

model_names_to_test = ["ann", "xgboost"]
results_data = []

print("Starting dataset processing using DataLoader...")

for dataset_id in CTR23:
    print(f"\n--- Processing Dataset ID: {dataset_id} ---")
    
    results_row = {'dataset_id': dataset_id}
    
    dataset = openml.datasets.get_dataset(dataset_id)
    X_raw, _, _, _ = dataset.get_data(target=dataset.default_target_attribute)
    results_row['name'] = dataset.name
    results_row['n_features_before_preprocessing'] = X_raw.shape[1]

    for model_name in model_names_to_test:
        print(f"  Training model: {model_name}...")
        
        X_train, y_train, X_test, y_test, model, prep = loader.load_from_openml(
            dataset_id=dataset_id,
            model_name=model_name,
            task_type="regression" 
        )

        if 'n_samples_train' not in results_row:
            results_row['n_samples_train'] = X_train.shape[0]
            results_row['n_samples_test'] = X_test.shape[0]
            results_row['n_features_after_preprocessing'] = X_train.shape[1]

        predictions = model.predict(X_test)
        mse = mean_squared_error(y_test, predictions)
        results_row[f'mse_{model_name}'] = mse
        print(f"    MSE for {model_name}: {mse:.4f}")

    results_data.append(results_row)

df_results = pd.DataFrame(results_data)

print("\n--- Final Results ---")
print(df_results)
```

```python
# save to csv
df_results.to_csv("CTR23_datasets_summary.csv", index=False)
```

# CC18

```python
import openml
import pandas as pd
import numpy as np
from bonXAI.core.utils import CC18
from bonXAI.core.data_loader import DataLoader
from sklearn.metrics import f1_score, roc_auc_score

loader = DataLoader()

model_names_to_test = ["ann", "xgboost"]
results_data = []

print("Starting classification dataset processing...")

for dataset_id in CC18:
    print(f"\n--- Processing Dataset ID: {dataset_id} ---")
    
    results_row = {'dataset_id': dataset_id}
    
    dataset = openml.datasets.get_dataset(dataset_id)
    X_raw, y_raw, _, _ = dataset.get_data(target=dataset.default_target_attribute)
    
    num_classes = len(np.unique(y_raw))
    is_multiclass = num_classes > 2
    
    results_row['name'] = dataset.name
    results_row['n_features_before_preprocessing'] = X_raw.shape[1]
    results_row['n_classes'] = num_classes

    for model_name in model_names_to_test:
        print(f"  Training model: {model_name}...")
        
        X_train, y_train, X_test, y_test, model, prep = loader.load_from_openml(
            dataset_id=dataset_id,
            model_name=model_name,
            task_type="classification" 
        )

        if 'n_samples_train' not in results_row:
            results_row['n_samples_train'] = X_train.shape[0]
            results_row['n_samples_test'] = X_test.shape[0]
            results_row['n_features_after_preprocessing'] = X_train.shape[1]

        y_pred = model.predict(X_test)
        y_pred_proba = model.predict_proba(X_test)
        
        if is_multiclass:
            f1 = f1_score(y_test, y_pred, average='weighted')
            roc_auc = roc_auc_score(y_test, y_pred_proba, multi_class='ovr', average='weighted')
        else:
            f1 = f1_score(y_test, y_pred)
            roc_auc = roc_auc_score(y_test, y_pred_proba[:, 1])

        results_row[f'f1_{model_name}'] = f1
        results_row[f'roc_auc_{model_name}'] = roc_auc
        print(f"    F1 Score for {model_name}: {f1:.4f}")
        print(f"    ROC AUC for {model_name}: {roc_auc:.4f}")

    results_data.append(results_row)

df_results = pd.DataFrame(results_data)

print("\n--- Final Results ---")
print(df_results)
```

```python
df_results.to_csv("CC18_datasets_summary.csv", index=False)
```
