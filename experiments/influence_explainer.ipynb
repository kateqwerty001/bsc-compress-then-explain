```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer
import sys
import os
notebook_dir = os.getcwd()
sys.path.append(os.path.abspath(os.path.join(notebook_dir, '..', 'custom_packages', 'bonXAI-main')))
from bonXAI.core.pytorch_ann import PyTorchANN
from bonXAI.core.explainer import Explainer  
```

```python
X, y = load_breast_cancer(return_X_y=True)
X = X.astype(np.float32)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)
```

```python
ann = PyTorchANN(
    task_type="classification",
    epochs=100,   
    lr=1e-3,
    batch_size=128,
    random_state=42
)
ann.fit(X_train, y_train, verbose=False)
```

```python
explainer = Explainer(
    model=ann,
    explainer_name="influence",
    strategy="influence",           
    task_type="classification" 
)
```

```python
X_fg = X_test[:10]
y_fg = y_test[:10]

influence_matrix, elapsed = explainer.explain(
    X_foreground=X_fg,
    X_background=X_train,
    n_jobs=1,
    y_foreground=y_fg,
    y_background=y_train
)
```
