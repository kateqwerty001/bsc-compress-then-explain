```python
from bonXAI.core.data_loader import DataLoader
from bonXAI.core.utils import set_global_seed
from bonXAI.core.utils import CC18_ALL, CTR23_ALL
```

```python
data_loader = DataLoader()
dataset_name, X_train, y_train, X_test, y_test, model, preprocessor = data_loader.load_from_openml(
    dataset_id=CC18_ALL[0], 
    model_name="ann", 
)
```

```python
from bonXAI.core.explainer import Explainer
exp = Explainer(model=model, explainer_name="shapiq", task_type="regression", seed=2)
x1 = exp.explain(X_foreground=X_test[:10], X_background=X_test[:1000], y_foreground=y_test[:10], y_background=y_test[:1000], n_jobs=10)
```

```python
from bonXAI.core.explainer import Explainer
exp = Explainer(model=model, explainer_name="shapiq", task_type="regression", seed=22)
x2 = exp.explain(X_foreground=X_test[:10], X_background=X_test[:1000], y_foreground=y_test[:10], y_background=y_test[:1000], n_jobs=10)
```

```python
from bonXAI.core.metrics import compute_mae

compute_mae(x1[0][0], x2[0][0])
```
