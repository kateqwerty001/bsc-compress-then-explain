```python
import os
```

```python
def check_ground_truth_status(dataset_name):
    root = "package_metadata/openml"
    target_path = os.path.join(root, dataset_name, "ground_truth")

    conditions = [
        ("SHAP Kernel ANN",    ["shap_", "kernel_", "ann"]),
        ("SAGE Perm ANN",      ["sage_", "permutation_", "ann"]),
        ("SHAP Kernel XGB",    ["shap_", "kernel_", "xgboost"]),
        ("SAGE Perm XGB",      ["sage_", "permutation_", "xgboost"]),
        ("SHAPIQ Kernel ANN",  ["shapiq_", "kernel_", "ann"]),
        ("SHAPIQ Kernel XGB",  ["shapiq_", "kernel_", "xgboost"]),
    ]

    if os.path.exists(target_path):
        files = [f.lower() for f in os.listdir(target_path)]
    else:
        files = []

    print(f"📊 check ground_truth for: {dataset_name}")
    print("-" * 30)

    results = []

    for label, keywords in conditions:
        is_found = False

        for fname in files:
            if all(word in fname for word in keywords):
                is_found = True
                break

        icon = "✅" if is_found else "❌"
        print(f"{icon} {label}")
        results.append(is_found)

    print("-" * 30 + "\n")
    return results
```

```python
def check_kernels_iid_status(dataset_name):
    root = "package_metadata/openml"
    target_path = os.path.join(root, dataset_name, "kernels_iid_comparison")

    conditions = [
        ("SHAP Kernel ANN",    ["shap_", "kernel_", "ann"]),
        ("SAGE Perm ANN",      ["sage_", "permutation_", "ann"]),
        ("SHAP Kernel XGB",    ["shap_", "kernel_", "xgboost"]),
        ("SAGE Perm XGB",      ["sage_", "permutation_", "xgboost"]),
        ("SHAPIQ Kernel ANN",  ["shapiq_", "kernel_", "ann"]),
        ("SHAPIQ Kernel XGB",  ["shapiq_", "kernel_", "xgboost"]),
    ]

    if os.path.exists(target_path):
        files = [f.lower() for f in os.listdir(target_path)]
    else:
        files = []

    print(f"📉 check kernels_iid_comparison for: {dataset_name}")
    print("-" * 50)

    results = []
    total_repeats = 0

    for label, keywords in conditions:
        found_files = []

        for fname in files:
            if all(word in fname for word in keywords):
                try:
                    repeat_str = fname.split('_')[-1]
                    repeat_str = repeat_str.split('.')[0]
                    repeats = int(repeat_str)
                except (ValueError, IndexError):
                    repeats = 0

                found_files.append((fname, repeats))

        if found_files:
            type_total = sum(repeats for _, repeats in found_files)
            total_repeats += type_total

            file_info = []
            for fname, repeats in found_files:
                file_info.append(f"{fname} ({repeats})")

            icon = "✅"
            status = f"found {len(found_files)} file(s), total repeats: {type_total}"
            details = f" [{', '.join(file_info)}]"
        else:
            icon = "❌"
            status = "not found"
            details = ""

        print(f"{icon} {label:25} {status:30} {details}")
        results.append(bool(found_files))

    print("-" * 50)
    print(f"📊 Total repeats across all methods: {total_repeats}")
    print()

    return results, total_repeats
```
