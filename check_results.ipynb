```python
import os

root = "package_metadata/openml"

columns = [
    "dataset",
    "ST SHAP ANN",
    "ST SAGE ANN",
    "ST SHAP XGB",
    "ST SAGE XGB",
    "ST SHAP-IQ ANN",
    "ST SHAP-IQ XGB"
]

patterns = {
    "ST SHAP ANN": ["st", "shap", "ann"],
    "ST SAGE ANN": ["st", "sage", "ann"],
    "ST SHAP XGB": ["st", "shap", "xgb"],
    "ST SAGE XGB": ["st", "sage", "xgb"],
    "ST SHAP-IQ ANN": ["st", "shapiq", "ann"],
    "ST SHAP-IQ XGB": ["st", "shapiq", "xgb"]
}

print("\t".join(columns))

missing = []   # <-- store all missing combinations

for dataset in os.listdir(root):
    dataset_path = os.path.join(root, dataset)
    if not os.path.isdir(dataset_path):
        continue

    other = os.path.join(dataset_path, "other_baselines")
    if not os.path.isdir(other):
        continue

    files = os.listdir(other)
    row = [dataset]

    for col in columns[1:]:
        needed = patterns[col]
        ok = 0

        for f in files:
            fname = f.lower()
            if all(word in fname for word in needed):
                ok = 1
                break

        row.append(str(ok))

        # record missing
        if ok == 0:
            missing.append((dataset, col))

    print("\t".join(row))

# ---- print missing at the end ----
print("\nMISSING DATA:")
for dataset, col in missing:
    print(f"{dataset} -> missing {col}")
```
