```python
import os

def list_files_in_dataset(dataset_name):
    root = "package_metadata/openml"
    subfolders_to_check = [
        "other_baselines",
        "kernels_iid_comparison",
        "ground_truth"
    ]

    dataset_dir = os.path.join(root, dataset_name)

    print(f"==========================================")
    print(f"📦 Dataset: {dataset_name}")


    for subfolder in subfolders_to_check:
        target_path = os.path.join(dataset_dir, subfolder)

        print(f"\n📂 Subfolder: {subfolder}")

        if not os.path.exists(target_path):
            print("   ❌  ")
            continue

        files = sorted(os.listdir(target_path))

        if not files:
            print("   ⚠️")
        else:
            for f in files:
                print(f"   - {f}")

    print(f"==========================================\n")
```

```python
import os

def check_ground_truth_status(dataset_name):
    root = "package_metadata/openml"
    target_path = os.path.join(root, dataset_name, "ground_truth")

    conditions = [
        ("SHAP Kernel ANN",    ["shap_", "kernel_", "ann"]),
        ("SAGE Perm ANN",      ["sage_", "permutation_", "ann"]),
        ("SHAP Kernel XGB",    ["shap_", "kernel_", "xgboost"]),
        ("SAGE Perm XGB",      ["sage_", "permutation_", "xgboost"]),
        ("SHAPIQ Kernel ANN",  ["shapiq_", "kernel_", "ann"]),
        ("SHAPIQ Kernel XGB",  ["shapiq_", "kernel_", "xgboost"]),
    ]

    if os.path.exists(target_path):
        files = [f.lower() for f in os.listdir(target_path)]
    else:
        files = []

    print(f"📊 check ground_truth for: {dataset_name}")
    print("-" * 30)

    results = []

    for label, keywords in conditions:
        is_found = False

        for fname in files:
            if all(word in fname for word in keywords):
                is_found = True
                break

        icon = "✅" if is_found else "❌"
        print(f"{icon} {label}")
        results.append(is_found)

    print("-" * 30 + "\n")
    return results
```
